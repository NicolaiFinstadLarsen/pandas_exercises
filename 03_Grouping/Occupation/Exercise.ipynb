```python
# Finne hvor mange det er av hvert gender i hver occuption. Fylle alle rader med transform. Count for å ikke telle NaN
# df.groupby(["occupation", "gender"])["gender"].transform("count")
df
```

```python
# # Finne hvor mange det er totalt gender i hver occuption. Fylle alle rader med transform. Count for å ikke telle NaN
# df.groupby(["occupation"])["gender"].transform("count")
df
```

```python
# Locate M gender i dataframe, hente ut kolonnene gender, occuptaion og ratio, droppe doble føringen grunnet transform, sortere høyest først.
by_gender = df.groupby(["occupation", "gender"])["gender"].transform("count")
total = df.groupby(["occupation"])["gender"].transform("count")
df["ratio_m_%"] = (by_gender / total) * 100
df.loc[
    df["gender"] == "M",
       ["occupation","ratio_m_%"]] \
        .drop_duplicates() \
            .sort_values(by="ratio_m_%", ascending=False)
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
df.groupby(["occupation"])["age"].agg(["min", "max"]).sort_values(by="max", ascending=False)
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
# Bruker named aggregation for penere oversikt da dette gir DataFrame og ikke series.
# Jeg synes dette er penere når jeg bygger en summary table
df.groupby(["occupation", "gender",]).agg(mean_age = ("age", "mean"))
```

### Step 8.  For each occupation present the percentage of women and men

```python
prct_df = df.groupby("occupation").agg(
    prct_women=("gender", lambda x: (x=='F').mean()*100),
    prct_male=("gender", lambda x: (x=='M').mean()*100),
).round(2)

prct_df["check_sum"] = prct_df["prct_women"] + prct_df["prct_male"]

# prct_df
prct_df.drop("check_sum", axis=1)
```

```python
(prct_df["prct_women"] + prct_df["prct_male"]).eq(100).all()
```
