### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv). 

### Step 3. Assign it to a variable called drinks.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv"
df = pd.read_csv(url)
```

```python
df.describe(include="all")
```

### Step 4. Which continent drinks more beer on average?

```python
df.groupby("continent")[["beer_servings"]].mean()
```

### Step 5. For each continent print the statistics for wine consumption.

```python
df.groupby("continent")[["wine_servings"]].describe()
```

### Step 6. Print the mean alcohol consumption per continent for every column

```python
df.groupby("continent").mean(numeric_only=True)
```

### Step 7. Print the median alcohol consumption per continent for every column

```python
df.groupby("continent").median(numeric_only=True)
```

### Step 8. Print the mean, min and max values for spirit consumption.
#### This time output a DataFrame

```python
# Dictionary aggregation
df.groupby("continent").agg(
    {"spirit_servings" : ["mean", "min", "max"],
    #  "wine_servings": ["mean", "min", "max"]
     })
```

```python
# Named aggregation
df.groupby("continent").agg(
    spirit_serving_mean = ("spirit_servings", "mean"),
    spirit_serving_min = ("spirit_servings", "min"),
    spirit_serving_max = ("spirit_servings", "max"),
)
```
