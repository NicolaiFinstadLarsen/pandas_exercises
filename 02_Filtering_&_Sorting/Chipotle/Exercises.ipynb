### Step 3. Assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
df = pd.read_csv(url, sep='\t')
```

```python
df.info()
```

### Step 4. How many products cost more than $10.00?

```python
df.dtypes
```

```python
df["item_price"] = df["item_price"].str.replace('$', '').astype(float)
```

```python
df["price_per_item"] = df["item_price"] / df["quantity"]
# df
```

```python
new_df = df.drop_duplicates(subset=["item_name", "choice_description"])
# new_df
```

```python
# Husk at man må sortere ut av et df for å kunne returnere et df
new_df_2 = new_df[new_df["price_per_item"] > 10]
```

```python
new_df_2
```

### Step 5. What is the price of each item? 
###### print a data frame with only three columns item_name choice_description and product_price

```python
new_df_2[["item_name", "choice_description", "price_per_item"]]
```

### Step 6. Sort by the name of the item

```python
new_df_2[
    ["item_name", "choice_description", "price_per_item"]
    ] \
    .sort_values(by="item_name", ascending=True)
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
# To me the question read as how many total items were sold of the item with the highest unit price.
# 

df[
    ["item_name", "quantity", "choice_description", "price_per_item"]
    ] \
    .sort_values(by="price_per_item", ascending=False)

df[df["item_name"] == "Steak Salad Bowl"]["quantity"].sum()
```

```python
# The question is actually, how many quantity were in the order with the highest price.
df.loc[df["item_price"].idxmax(), "quantity"]
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
df.loc[df["item_name"] == "Veggie Salad Bowl", "quantity"].sum() #recommended
# df[df["item_name"] == "Veggie Salad Bowl"]["quantity"].sum()
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
df[(df["item_name"] == "Canned Soda") & \
   (df["quantity"] >= 2)].shape[0]
```
